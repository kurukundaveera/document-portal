# Read the Data

# Now Will do the Chunking 
# Chuncking strategy is only possible for text data , not for audio , video or image data.
## this is a experimental thing there is no deterministic way to split the text

```python
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=150,
    length_function=len
)
```

```python
docs=text_splitter.split_documents(documnets)
```

```python
len(docs)
```

# Now store the chunked data in the vector database(FAISS- InMemory Storage , which means we are stroing in the RAM itself)

```python
from langchain.vectorstores import FAISS
```

```python
docs[0].page_content
```

```python
embedding_model.embed_documents(docs[0].page_content)
```

# Size of the above Vector

```python
len(embedding_model.embed_documents(docs[0].page_content)[0])
```

```python
vectorstore=FAISS.from_documents(docs, embedding_model)
```


# Question: this is a user question
# context: based on the question retrieving the info from the vector  database


```python
prompt_template = """
        Answer the question based on the context provided below. 
        If the context does not contain sufficient information, respond with: 
        "I do not have enough information about this."

        Context: {context}

        Question: {question}

        Answer:"""
```

# Create object for the prompt template

```python
from langchain.prompts import PromptTemplate
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variable=["context", "question"]
)
```

# This prompt template will pass to the LLM

```python
prompt
```

# Now Create a chaining LCEL (langchain expression langauage)Concept
# what ever outside generate by the LLM/model will pass it to the output paser

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser=StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
rag_chain.invoke("can you tell me the granular reward model acuuracy per preference rating from table8")
```

```python
rag_chain.invoke("tell  me about the llama2 finetuning benchmark experiments?")
```
